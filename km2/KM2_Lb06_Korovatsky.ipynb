## Задание 6.1.3. Функция flatten_list: создание, документирование, тестирование

```python
def flatten_list(nested_obj: list, gen: bool = False):
    
    def flatten_list_v1(nested_obj):
        result = []
        if isinstance(nested_obj, list):
            for item in nested_obj:
                result += flatten_list_v1(item)
        else:
            result += [nested_obj]
        return result

    def flatten_list_v2b(nested_obj):
        if isinstance(nested_obj, list):
            for item in nested_obj:
                yield from flatten_list_v2b(item)
        else:
            yield nested_obj

    genDict = {"True": flatten_list_v2b,
              "False": flatten_list_v1}
    return genDict[str(gen)](nested_obj)
```

## Задание 6.2.1. Рекурсивная функция flatten_it_v1

```python
def flatten_it_v1(nested_obj):
    result = []
    if hasattr(nested_obj, '__iter__') and not isinstance(nested_obj, str):
        if isinstance(nested_obj, dict):
            for DItem in nested_obj.items():
                result += flatten_it_v1(DItem)
        else:
            for item in nested_obj:
                result += flatten_it_v1(item)
    else:
        result += [nested_obj]
    return result
```

## Задание 6.2.2. Генераторная функция на основе рекурсии flatten_it_v2

```python
def flatten_it_v2(nested_obj):
    if hasattr(nested_obj, '__iter__') and not isinstance(nested_obj, str):
        if isinstance(nested_obj, dict):
            for DItem in nested_obj.items():
                yield from flatten_it_v2(DItem)
        else:
            for item in nested_obj:
                yield from flatten_it_v2(item)
    else:
        yield nested_obj
```

```python
def flatten_list_v1(nested_obj):
    result = []
    if isinstance(nested_obj, list):
        for item in nested_obj:
            if item is nested_obj:
                raise ValueError("Cyclic list is found")
            else:
                result += flatten_list_v1(item)
    else:
        result += [nested_obj]
    return result
```

## Задание 6.3.1 и 6.3.2

```python
def flatten_it_fin(nested_obj, gen: bool = False):

    def flatten_it_v1(nestedObj, firstLevel = None):
        if firstLevel is None:
            firstLevel = nestedObj
        result = []
        
        if hasattr(nestedObj, "__iter__") and not isinstance(nestedObj, str):
            if isinstance(nestedObj, dict):
                for key, value in nestedObj.items():
                    if value is firstLevel: 
                        raise ValueError("Cyclic object is found")
                    result += flatten_it_v1((key, value), firstLevel)
            else:
                for item in nestedObj:
                    if item is firstLevel:
                        raise ValueError("Cyclic object is found")
                    result += flatten_it_v1(item, firstLevel)
        else:
            result += [nestedObj]
        
        return result


    def flatten_it_v2(nestedObj):
        if hasattr(nestedObj, "__iter__") and not isinstance(nestedObj, str):
            if isinstance(nestedObj, dict):
                for dictItem in nestedObj.items():
                    yield from flatten_it_v2(dictItem)
            else:
                for item in nestedObj:
                    if item is nestedObj:
                        raise ValueError("Cyclic object is found")
                    yield from flatten_it_v2(item)
        else:
            yield nestedObj

    
    genDict = {"True": flatten_it_v2,
              "False": flatten_it_v1}
    return genDict[str(gen)](nested_obj)
```

```python
testCyclic = {'a' : 1}
testCyclic['b'] = {2 : testCyclic}

try:
    print(flatten_it_fin(testCyclic))
except ValueError as text:
    print(text)
```
