## Задание 7.4 Перегрузка операции принадлежности in

```python
# переопределение классов
class BinaryTree:
    def __init__(self):
        self.root = EmptyNode()
    def __repr__(self):
        return repr(self.root)
    def insert(self, value):
        self.root = self.root.insert(value)
    def __contains__(self, value):
        return value in self.root

class EmptyNode:
    def __repr__(self):
        return '*'
    def insert(self, value):
        return BinaryNode(self, value, self)
    def __contains__(self, value):
        return False

class BinaryNode:
    def __init__(self, left, value, right):
        self.left = left
        self.value = value
        self.right = right
    def __repr__(self):
        return f'({self.left}, {self.value}, {self.right})'
    def insert(self, value):
        if value < self.value:
            if isinstance(self.left, EmptyNode):
                self.left = BinaryNode(EmptyNode(), value, EmptyNode())
            else:
                self.left.insert(value)
                
        elif value > self.value:
            if isinstance(self.right, EmptyNode):
                self.right = BinaryNode(EmptyNode(), value, EmptyNode())
            else:
                self.right.insert(value)
        return self
    def __contains__(self, value):
        if self == value:
            return True
        elif value < self.value:
            return self.left.__contains__(value)
        else:  # value > self.value
            return self.right.__contains__(value)
```

```python
sourse_data = [5,1,10,3,4]
treeN = BinaryTree()
for i in sourse_data:
    treeN.insert(i)
    print(treeN)

for i in range(10):
    print((i, treeN.__contains__(i)), end = ' ')
```

## Задание 7.5 Перегрузка встроенной функции len

```python
# переопределение классов 2
class BinaryTree:
    def __init__(self):
        self.root = EmptyNode()
    def __repr__(self):
        return repr(self.root)
    def insert(self, value):
        self.root = self.root.insert(value)
    def __contains__(self, value):
        return value in self.root
    def __len__(self):
        return len(self.root) - 1

class EmptyNode:
    def __repr__(self):
        return '*'
    def insert(self, value):
        return BinaryNode(self, value, self)
    def __contains__(self, value):
        return False
    def __len__(self):
        return 1

class BinaryNode:
    def __init__(self, left, value, right):
        self.left = left
        self.value = value
        self.right = right
    def __repr__(self):
        return f'({self.left}, {self.value}, {self.right})'
    def insert(self, value):
        if value < self.value:
            if isinstance(self.left, EmptyNode):
                self.left = BinaryNode(EmptyNode(), value, EmptyNode())
            else:
                self.left.insert(value)
                
        elif value > self.value:
            if isinstance(self.right, EmptyNode):
                self.right = BinaryNode(EmptyNode(), value, EmptyNode())
            else:
                self.right.insert(value)
        return self
    def __contains__(self, value):
        if self == value:
            return True
        elif value < self.value:
            return self.left.__contains__(value)
        else:  # value > self.value
            return self.right.__contains__(value)
    def __len__(self):
        return len(self.left) + len(self.right)
```

```python
sourse_data = [5,1,10,3,4]
treeLen = BinaryTree()
for i in sourse_data:
    treeLen.insert(i)
print(treeLen)
len(treeLen)
```
